```python
import pandas as pd
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer

nltk.download('vader_lexicon')

df = pd.read_csv('/content/sample_data/ts_discography_released.csv')

df = df.drop_duplicates(subset='song_title')

def preprocess_text(text):
    text = text.replace('\n', ' ')
    text = ''.join([char for char in text if char.isalnum() or char.isspace()])
    text = text.lower()
    return text

df['song_lyrics'] = df['song_lyrics'].apply(preprocess_text)

sid = SentimentIntensityAnalyzer()

def analyze_sentiment(text):
    sentiment_score = sid.polarity_scores(text)
    return sentiment_score

df['sentiment'] = df['song_lyrics'].apply(analyze_sentiment)

df.to_csv('taylor_swift_sentiment_analysis.csv', index=False)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/sample_data/taylor_swift_spotify.csv', index_col='id', parse_dates=['release_date'])
df.head()
df.drop(columns=['Unnamed: 0', 'uri'], inplace=True)
df.album.value_counts()
df = df[~(df['album'] == 'Live From Clear Channel Stripped 2008')]

df_1 = df.copy()
df_2 = df.copy()
```

```python
def clean_album(album):
  if album in ['folklore: the long pond studio sessions (from the Disney+ special) [deluxe edition]', 'folklore (deluxe version)', 'folklore']:
    return 'folklore'
  elif album in ['reputation Stadium Tour Surprise Song Playlist', 'reputation']:
    return 'reputation'
  elif album in ['THE TORTURED POETS DEPARTMENT: THE ANTHOLOGY', 'THE TORTURED POETS DEPARTMENT']:
    return 'THE TORTURED POETS DEPARTMENT'
  elif album in ["Red (Taylor's Version)", 'Red (Deluxe Edition)', 'Red']:
    return 'Red'
  elif album in ["Fearless (Taylor's Version)", 'Fearless Platinum Edition', 'Fearless']:
    return 'Fearless'
  elif album in ['Midnights (The Til Dawn Edition)', 'Midnights (3am Edition)', 'Midnights']:
    return 'Midnights'
  elif album in ["Speak Now (Taylor's Version)", 'Speak Now (Deluxe Edition)', 'Speak Now World Tour Live', 'Speak Now']:
    return 'Speak Now'
  elif album in ["1989 (Taylor's Version) [Deluxe]", "1989 (Taylor's Version)", '1989 (Deluxe Edition)', '1989']:
    return '1989'
  elif album in ['evermore (deluxe version)', 'evermore']:
    return 'evermore'
  else:
    return album

df_1['album'] = df_1['album'].apply(lambda x: clean_album(x))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define the album colors
album_colors = {
    'Speak Now': '#813c60',
    '1989': '#cdc6ac',
    'Fearless': '#865829',
    'evermore': '#7e5c43',
    'folklore': '#7f7f7f',
    'reputation': '#333333',
    'Taylor Swift': '#50a7e0',
    'Lover': '#f7cacd',
    'Red': '#a02b48',
    'Midnights': '#a3a277',
    'THE TORTURED POETS DEPARTMENT': '#000000'
}

fig = plt.figure(figsize=(10, 7))
ax = plt.axes()

# Iterate through the unique albums
for album in df_1['album'].unique():
    # Filter data for the current album
    album_data = df_1[df_1['album'] == album]

    # Create a bar plot for the current album
    sns.barplot(data=album_data, x='acousticness', y='album', ci=None, ax=ax, color=album_colors[album])

    # Calculate the mean acousticness for the current album
    mean_acousticness = album_data['acousticness'].mean()

    # Plot a central line for the mean value
    ax.axvline(x=mean_acousticness, color='black', linestyle='--', linewidth=1)

ax.set_title('Acousticness by Album')
ax.set_xlabel('Acousticness')
ax.set_ylabel('Album')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Assuming df_1 is already defined and contains 'album' and 'acousticness' columns

# Aggregate the data to find the mean 'acousticness' for each album
agg_data = df_1.groupby('album')['acousticness'].mean().reset_index()

fig = plt.figure(figsize=(10, 6))
ax = plt.axes()

# Create a bar chart
barplot = sns.barplot(data=agg_data, x='acousticness', y='album', palette=album_colors, ax=ax)

# Annotate each bar with the corresponding mean value
for index, row in agg_data.iterrows():
    ax.text(row['acousticness'], index, f"{row['acousticness']:.2f}", color='black', ha="left", va="center")

plt.show()
```
